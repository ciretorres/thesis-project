```python
# https://pandas.pydata.org/
import pandas

# https://astroquery.readthedocs.io/en/latest/
from astroquery.vizier import Vizier
```

# Vizier

```python
# https://github.com/cds-astro/cds.cdsclient
# clear
Vizier.clear_cache()

columns = ['recno', 'HIP', 'Proxy', 'RAhms', 'DEdms', 'Vmag', 'VarFlag', 'r_Vmag',
       'RAICRS', 'DEICRS', 'AstroRef', 'Plx', 'pmRA', 'pmDE', 'e_RAICRS',
       'e_DEICRS', 'e_Plx', 'e_pmRA', 'e_pmDE', 'DE:RA', 'Plx:RA', 'Plx:DE',
       'pmRA:RA', 'pmRA:DE', 'pmRA:Plx', 'pmDE:RA', 'pmDE:DE', 'pmDE:Plx',
       'pmDE:pmRA', 'F1', 'F2', 'BTmag', 'e_BTmag', 'VTmag', 'e_VTmag',
       'm_BTmag', 'B-V', 'e_B-V', 'r_B-V', 'V-I', 'e_V-I', 'r_V-I', 'CombMag',
       'Hpmag', 'e_Hpmag', 'Hpscat', 'o_Hpmag', 'm_Hpmag', 'Hpmax', 'HPmin',
       'Period', 'HvarType', 'moreVar', 'morePhoto', 'CCDM', 'n_CCDM', 'Nsys',
       'Ncomp', 'MultFlag', 'Source', 'Qual', 'm_HIP', 'theta', 'rho', 'e_rho',
       'dHp', 'e_dHp', 'Survey', 'Chart', 'Notes', 'HD', 'BD', 'CoD', 'CPD',
       '(V-I)red', 'SpType', 'r_SpType', 'HIPep', 'Erratum', '_RA.icrs',
       '_DE.icrs']

row_limit = -1  # remove row limit if you want everything

# Create an instance of the Vizier class
#vizier = Vizier()
vizier = Vizier( 
    columns=columns, 
    vizier_server='vizier.cds.unistra.fr', 
    row_limit=row_limit
)

# Retrieve the Hipparcos catalogue
hipparcos_catalog = vizier.get_catalogs( catalog="I/239/hip_main" )
hip_cat = hipparcos_catalog[0]

# "watch" / preview

print(hipparcos_catalog)

print(len(hip_cat.colnames)) # columns

print(len(hip_cat))          # number of rows

print(hip_cat)               # quick preview
```
