## Removiendo las malas entradas

```python
# dataframe en donde los valores de Plx sean mayores a 0
# y en Rahms, DEdms y Vmag no sean NaN
# De 118,218 se reduce a 113,710

# remove bad entries and higher error
valid_mask = (
    ( df['Plx'] > 0) &
    
    ( df['Vmag'].notna() )  &
    
    ( df['RAICRS'].notna() ) &
    ( df['DEICRS'].notna() ) &
    ( df['RAhms'].notna() ) &
    ( df['DEdms'].notna() )  
)

df_nan_par_corrected = df[ valid_mask ].copy()

df_nan_par_corrected
```
